Question 6: Handle Missing Values

a. Drop rows based on column

```python
df = df.dropna(subset=['sales']).reset_index(drop=True) # droping rows will change indices thus it is necessary
print(df.isnull().sum())
```

b. Imputing

```python
df["profit"]=df["profit"].fillna(df["profit"].mean())
df["discount"]=df["discount"].fillna(df["discount"].median())
print(df.isnull().sum())
```

**Question 7: Handling missing categorical data**

```python
# First, check what columns are available in your DataFrame
print("Available columns:", df.columns.tolist())

# Check if the column exists before trying to fill NaN values
if 'customer_name' in df.columns:
    df["customer_name"] = df["customer_name"].fillna("Unknown")
else:
    # If the exact column name doesn't exist, you might need to find the correct name
    # Look for similar column names (case-insensitive search)
    possible_columns = [col for col in df.columns if 'customer' in col.lower() and 'name' in col.lower()]
    if possible_columns:
        print(f"Found similar columns: {possible_columns}")
        # Use the first matching column (adjust as needed)
        df[possible_columns[0]] = df[possible_columns[0]].fillna("Unknown")
    else:
        print("No customer name column found. Please check your DataFrame structure.")

# Check for null values after the operation
print(df.isnull().sum())
```

```python
duplicate_values = df.duplicated(subset=["order_id"])

print("Number of duplicated values:", duplicate_values.sum())

before_removal = len(df)

df = df.drop_duplicates(subset=["order_id"])

duplicate_count = df.duplicated(subset=["order_id"]).sum()

after_removing = len(df)

df.reset_index(drop=True, inplace=True)

print("Number of duplicate values removed: ", before_removal-after_removing)

print("Number of duplicates remaining:", duplicate_count)
```
